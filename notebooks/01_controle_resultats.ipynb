```python
import pandas as pd

matchs = pd.read_csv("../data/resultats_ligue1.csv", parse_dates=["Date"])

# Valeurs manquantes par colonne
display(matchs.isna().sum())

# Clubs présents et nombre de saisons disputées
domicile = matchs[["HomeTeam", "Saison"]].rename(columns={"HomeTeam": "Club"})
exterieur = matchs[["AwayTeam", "Saison"]].rename(columns={"AwayTeam": "Club"})
clubs = pd.concat([domicile, exterieur]).drop_duplicates()
display(clubs.groupby("Club").size().sort_values(ascending=False))
```
