```python
import duckdb
con = duckdb.connect("../data/transfermarkt.duckdb", read_only=True)
# Liste des tables et de leurs colonnes
tables = con.sql("SHOW ALL TABLES").df()
display(tables[["name", "column_names"]])
```

```python
for nom in ["clubs", "games", "player_valuations", "transfers", "game_lineups"]:
    colonnes = tables.loc[tables["name"] == nom, "column_names"].iloc[0]
    print(nom, ":", list(colonnes), "\n")
```

```python
# Clubs de Ligue 1 (code FR1) depuis la saison 2021-2022
clubs_tm = con.sql("""
    SELECT DISTINCT home_club_id AS club_id, home_club_name AS nom_tm
    FROM games
    WHERE competition_id = 'FR1' AND CAST(season AS INTEGER) >= 2021
    ORDER BY nom_tm
""").df()
display(clubs_tm)

# Contrôle : nombre de matchs par saison
display(con.sql("""
    SELECT season, COUNT(*) AS nb_matchs
    FROM games
    WHERE competition_id = 'FR1' AND CAST(season AS INTEGER) >= 2021
    GROUP BY season
    ORDER BY season
""").df())
```
